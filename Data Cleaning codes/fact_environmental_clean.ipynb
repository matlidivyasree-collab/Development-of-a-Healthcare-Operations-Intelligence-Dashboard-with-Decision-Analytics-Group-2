```python
df["aqi"]=df["aqi"].fillna(df["aqi"].median())
```

```python
df["water_quality_index"]=df["water_quality_index"].fillna(df["water_quality_index"].median())
```

```python
df["sanitation_coverage_pct"]=df["sanitation_coverage_pct"].fillna(df["sanitation_coverage_pct"].median())
```

```python
df["hotspot_flag"]=df["hotspot_flag"].astype(str).str.lower().str.strip()
df["hotspot_flag"]=df["hotspot_flag"].replace({'yes':'true','1':'true','y':'true','no':'false','0':'false','n':'false'})
df["hotspot_flag"].value_counts()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.to_csv(r"C:\Users\attil\Downloads\fact_environmental_cleaned.csv",index=False)
```
