```python
df["difference"] = (
    df["total_reported_cases"]
    - (
        df["active_cases"]
        + df["recovered_cases"]
        + df["deaths"]
    )
)

df["difference"].describe()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df["hospitalized_cases"]=df["hospitalized_cases"].replace("-",np.nan)
```

```python
df["hospitalized_cases"]=pd.to_numeric(
    df["hospitalized_cases"],
    errors="coerce"
)
```

```python
df["hospitalized_cases"].dtype
```

```python
df.loc[df["active_cases"]<0, "active_cases"]=np.nan
```

```python
df.loc[df["deaths"]<0,"deaths"]=np.nan
```

```python
cols = [
    "active_cases",
    "recovered_cases",
    "deaths",
    "hospitalized_cases",
    "icu_admissions",
    "case_fatality_rate"
]

for col in cols:
    df[col] = df[col].fillna(df[col].median())
```
